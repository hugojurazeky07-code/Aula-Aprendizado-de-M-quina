```python
# Importar a blibioteca pandas
import pandas as pd

# Dicionario (simulando um dataframe)
dados = {
 "Nome": ["Ana","Carlos","João","Maria"],
 "Idade": [20,22,19,25],
 "Nota": [8.5, 7.0, 9.0, 6.5]
}

# Converter para a estrutura principal (Dataframe)
df = pd.DataFrame(dados)

# Exibir o conteudo
print(df)
```

```python
# Exibir somente a coluna Nome
print(df["Nome"])
print()

# Exibir as colunas Nome e Nota
print(df[["Nome", "Nota"]])
```

```python
# Calculos nos dados
print("Média: ", df["Nota"].mean())

# Maior nota
print("Maior nota: ", df["Nota"].max())

# Menor nota
print("Menor nota: ", df["Nota"].min())
```

```python
# Filtros de dados
aprovados = df[df["Nota"] >= 7]
reprovados = df[df["Nota"] < 7]

print(aprovados)
print()
print(reprovados)
```

```python
# Filtar com 2 condições
resultado =df[(df["Nota"] >= 7) & (df["Idade"] >= 20)]
print(resultado)
```

```python
# Adicionar uma nova coluna
df["Situacao"] = "Reprovado"
print(df)
print()

# Alterar um valor na nova coluna
df.loc[df["Nota"] >= 7, "Situacao"] = "Aprovado"
print(df)
```
